```python
# import the necessary libraries
from qiskit import QuantumCircuit
from qiskit.visualization import array_to_latex

from qiskit_aer import Aer
```

```python
# create a quantum circuit with 3 qubits
qc = QuantumCircuit(3)
```

```python
# qc.x(1)  # Apply X gate to qubit 1
# qc.h(0)  # Apply H gate to qubit 0

qc.ccx(0, 1, 2)  # Toffoli gate (CCX gate)
qc.draw(output='mpl')
```

```python
# using Unitary simulator, simulate the unitary matrix of the circuit
backend = Aer.get_backend('unitary_simulator')
output = backend.run(qc).result().get_unitary()

# Display the matrix in LaTeX
display(array_to_latex(output, prefix="\\text{Circuit output = }"))
```

```python
# using Statevector simulator, simulate the statevector of the circuit
backend = Aer.get_backend('statevector_simulator')
output = backend.run(qc).result().get_statevector()

# Display the statevector in LaTeX
display(array_to_latex(output, prefix="\\text{Statevector} = "))
```
